```python
import os, glob
import pandas as pd
```

```python
files = glob.glob("../../data/regression_outputs/regmodels/Fold/*/*/Fuse/Multi_Concat/results.csv")
len(files)
```

```python
label_sdg_files = glob.glob("../../data/processed/0labels/*.csv")
labels_sdg_list = []
for label_sdg_file in label_sdg_files:
    labels_sdg = pd.read_csv(label_sdg_file)
    labels_sdg_list.append(labels_sdg)
labels_sdg = pd.concat(labels_sdg_list, ignore_index=True)
labels_sdg
```

```python
df_ls = []
for file in files:
    city_tmp = pd.read_csv(file)
    city_tmp['country'] = file.split('/')[-5]
    city_tmp['city'] = file.split('/')[-4]
    df_ls.append(city_tmp)
df = pd.concat(df_ls, ignore_index=True)
df
```

```python
df = df[df['r2'] > 0]
df = df[df['target'].isin(labels_sdg['ID'].unique())]
df.reset_index(drop=True, inplace=True)
df
```

```python
df['country_city'] = df['country'] + '_' + df['city']
del df['country'], df['city'], df['target'], df['fold']
df
```

```python
city_df = df.groupby(['country_city']).mean().reset_index()
city_df
```

```python
city_center = pd.read_csv('../../data/processed/all_center.csv')
city_center['country_city'] = city_center['country'] + '_' + city_center['city']
city_center
```

```python
city_df = pd.merge(city_df, city_center[['country_city', 'lat', 'lon']], on='country_city', how='left')
city_df
```

```python
city_df.to_csv("../../data/processed/fig/fig1c_map_spearman.csv", index=False)
```
