```python
import os, glob
import pandas as pd
```

```python
label_sdg_files = glob.glob("../../data/processed/0labels/*.csv")
labels_sdg_list = []
for label_sdg_file in label_sdg_files:
    labels_sdg = pd.read_csv(label_sdg_file)
    labels_sdg_list.append(labels_sdg)
labels_sdg = pd.concat(labels_sdg_list, ignore_index=True)
labels_sdg
```

```python
def load_data(path_pattern):
    files = glob.glob(path_pattern)
    print(len(files))
    df_ls = []
    for file in files:
        city_tmp = pd.read_csv(file)
        city_tmp['city'] = file.split('/')[-4]
        df_ls.append(city_tmp)
    df = pd.concat(df_ls, ignore_index=True)
    return df
```

```python
sv_df = load_data("../../data/regression_outputs/regmodels_spatial_self/Fold/*/*/SV/Token_Concat_self_spatial/results.csv")
rs_df = load_data("../../data/regression_outputs/regmodels_spatial_self/Fold/*/*/RS/Token_Concat_self_spatial/results.csv")
both_df = load_data("../../data/regression_outputs/regmodels_spatial_self/Fold/*/*/Fuse/Token_Concat_spatial_self/results.csv")
```

```python
all_df = sv_df.merge(rs_df, on=['city', 'target'], suffixes=('_sv', '_rs'))
all_df = all_df.merge(both_df, on=['city', 'target'])
all_df
```

```python
all_df = all_df[all_df['target'].isin(labels_sdg['ID'].unique())]
all_df.reset_index(drop=True, inplace=True)
all_df
```

```python
all_df['SDG'] = all_df['target'].map(dict(zip(labels_sdg['ID'], labels_sdg['SDG'])))
all_df
```

```python
output_file = "../../data/processed/fig/fig2_modal_compare.csv"
all_df.to_csv(output_file, index=False)
```
