```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from matplotlib import font_manager
font_path = Path('../../data/Arial.ttf')
if font_path.exists():
    font_manager.fontManager.addfont(str(font_path))
    plt.rcParams['font.family'] = 'Arial'
else:
    plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['pdf.fonttype'] = 42
plt.rcParams['ps.fonttype'] = 42
```

```python
# Final manuscript Fig. 3b shows all four satellite concepts reported in the Supplement.
rs_order = ['concrete', 'rooftop', 'vegetation', 'soil']
rs_concept_df = pd.read_csv('../../data/processed/fig/clip_concept_rs.csv')
rs_concept_df['concept'] = rs_concept_df['concept'].str.lower()
rs_concept_df = rs_concept_df.set_index('concept').loc[rs_order].reset_index()
rs_concept_df['concept'] = rs_concept_df['concept'].str.capitalize()
rs_concept_df.columns = ['concept', 'RS-self', 'RS-spatial']
rs_concept_df
```

```python
colors = ['#FFCA28', '#FFE082']
fig, ax = plt.subplots(figsize=(6, 4))
df_melted = rs_concept_df.melt(id_vars='concept', value_vars=['RS-self', 'RS-spatial'], var_name='metric', value_name='r2_value')
sns.barplot(data=df_melted, x='concept', y='r2_value', hue='metric', ax=ax, palette=colors, width=0.5)
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)
ax.tick_params(axis='both', which='major', labelsize=12)
ax.legend(title='', fontsize=12, frameon=False)
ax.set_xlabel('')
ax.set_ylabel('$R^2$', fontsize=14)
plt.tight_layout()
plt.savefig('../../data/figure_assets/fig2_clip_concept_rs.svg')
plt.show()
```
